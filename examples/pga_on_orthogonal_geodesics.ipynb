# Principal geodesic analysis on orthogonal geodesics

```python
import os
import sys
from pathlib import Path
sys.path.append(str(Path(os.getcwd()).parent.absolute()))
```

```python
import numpy as np

from pca.bures_wasserstein_principal_geodesic_analysis import BuresWassersteinPGA, BuresWassersteinTPCA
from tools.generate import generate_spd_matrices_on_orthogonal_geodesics
from tools.plot_2d import plot_results
```

```python
dim = 2
n_points = 20
end_times = [0.7, 0.5, 0.3]
times = np.vstack([np.linspace(-t, t, n_points) for t in end_times])
points_spd, mean_spd = generate_spd_matrices_on_orthogonal_geodesics(dim, times, seed=1)
pga = BuresWassersteinPGA(dim).fit(points_spd)
```

```python
plot_results(points_spd, pga)
```

```python
times = np.vstack([
    np.linspace(-1., 3., n_points),
    np.linspace(-1., 1., n_points),
    np.linspace(-0.5, 0.5, n_points),
])
points_spd, mean_spd = generate_spd_matrices_on_orthogonal_geodesics(dim, times, seed=1)
pga = BuresWassersteinPGA(dim).fit(points_spd)
plot_results(points_spd, pga)
```

```python
tpca = BuresWassersteinTPCA.fit(points_spd)
plot_results(points_spd, tpca)
```
