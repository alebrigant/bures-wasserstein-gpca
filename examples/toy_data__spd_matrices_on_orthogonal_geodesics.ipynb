# GPCA and TPCA on orthogonal geodesics

In this notebook, we perform GPCA and TPCA on a set of $2\times 2$ SPD matrices obtained by sampling from 3 orthogonal geodesics that intersect at their common midpoint. We can see that for both methods, those 3 orthogonal geodesics are retrieved by the geodesic components, in order of decreasing lenght.

```python
import os
import sys
from pathlib import Path
sys.path.append(str(Path(os.getcwd()).parent.absolute()))
```

```python
import numpy as np

from pca.bures_wasserstein_geodesic_pca import *
from pca.bures_wasserstein_tangent_pca import *
from tools.plot_2d import *
from tools.generate import *
```

#### Sample SPD matrices on 3 orthogonal Bures-Wasserstein geodesics

```python
dim = 2
n_points = 21
points_spd, mean_spd = generate_spd_matrices_on_orthogonal_geodesics(dim, n_points, seed=1)
```

#### Apply GPCA and visualize the projections on each component

```python
gpca = BuresWassersteinGPCA(dim).fit(points_spd)
print(gpca.costs)
plot_pca_results_with_ellipses(points_spd, gpca)
```

#### Apply TPCA and visualize the projections on each component

```python
tpca = BuresWassersteinTPCA().fit(points_spd)
print(tpca.costs)
plot_pca_results_with_ellipses(points_spd, tpca)
```
